# Zomato’s Restaurant Data Anlysis


Understanding customer preferences and restaurant trends is important for making informed business decisions in food industry. In this project, we will analyze Zomato’s restaurant dataset to find meaningful insights. We aim to answer questions such as:

- Do more restaurants provide online delivery compared to offline services?
- Which types of restaurants are most favored by the general public?
- What price range do couples prefer for dining out?

#### Step 1: Importing necessary Python libraries.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

#### Step 2: Reading the data

```python
dataframe = pd.read_csv("Zomato-data-.csv")
print(dataframe.head())
```

#### Step 3: Data Cleaning and Preparation

Before moving further we need to clean and process the data, we need to convert the rate column to a float by removing denominator characters - applies the handleRate function to clean and convert each rating value in the 'rate' column.

```python
def handleRate(value):
    value=str(value).split('/')[0]
    return float(value)

dataframe['rate']=dataframe['rate'].apply(handleRate)
print(dataframe.head())
```

Next, let us check all null values

```python
dataframe.info()
```

#### Step 4: Exploring Restaurant Types

```python
plt.figure(figsize=(5,3))
sns.countplot(x=dataframe['listed_in(type)'])
plt.xlabel("Type of restairant")
```

Conclusion: The majority of the restaurants fall into the dining category

```python
vote_grouped = dataframe.groupby('listed_in(type)')['votes'].sum()
result = pd.DataFrame({'votes': vote_grouped})

plt.figure(figsize=(5,3))
plt.plot(result, c='green', marker='o')
plt.xlabel('Type of restaurant')
plt.ylabel('Votes')
```

Conclusion: Dining restaurants are preferred by a larger number of individuals.

#### Step 5: Identify the Most Voted Restaurant

```python
max_votes = dataframe['votes'].max()
restaurant_with_max_votes = dataframe.loc[dataframe['votes'] == max_votes, 'name']

print('Restaurant(s) with the maximum votes:', restaurant_with_max_votes)
```

Now we find the restaurant with the highest number of votes is Empire Restaurant

#### Step 6: Online Order Availability

```python
plt.figure(figsize=(5,3))
sns.countplot(x=dataframe['online_order'])
plt.xlabel('Restaurants accept online orders (Y/N)')
```

Conclusion: This suggests that a majority of the restaurants do not accept online orders.

#### Step 7: Analyze Ratings

```python
plt.figure(figsize=(5,3))
plt.hist(dataframe['rate'],bins =5)
plt.title('Ratings Distribution')
plt.show()
```

Conclusion: The majority of restaurants received ratings ranging from 3.5 to 4.

#### Step 8: Approximate Cost for Couples

```python
couple_data=dataframe['approx_cost(for two people)']
plt.figure(figsize=(7,3))
sns.countplot(x=couple_data)
```

Conclusion: The majority of couples prefer restaurants with an approximate cost of 300 rupees.

#### Step 9: Ratings Comparison - Online vs Offline Orders

```python
plt.figure(figsize=(3,3))
sns.boxplot(x='online_order', y='rate', data=dataframe)
```

Conclusion: Offline orders received lower ratings in comparison to online orders which obtained excellent ratings.

#### Step 10: Order Mode Preferences by Restaurant Type

```python
pivot_table = dataframe.pivot_table(index='listed_in(type)', columns='online_order', aggfunc='size', fill_value=0)
plt.figure(figsize=(5,4))
sns.heatmap(pivot_table, annot=True, cmap='YlGnBu', fmt='d')
plt.title('Heatmap')
plt.xlabel('Online Order')
plt.ylabel('Listed In (Type)')
plt.show()
```

With this we can say that dining restaurants primarily accept offline orders whereas cafes primarily receive online orders. This suggests that clients prefer to place orders in person at restaurants but prefer online ordering at cafes
